```python
import numpy as np
import pandas as pd
from sklearn.linear_model import Perceptron
from mlxtend.plotting import plot_decision_regions
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('binary_classification_dataset.csv')
```

```python
df.head()
```

```python
df.drop(['feature_1','feature_2','feature_3'],axis=1,inplace=True)
```

```python
sns.scatterplot(data=df,x='feature_4',y='feature_5',hue='target')
```

```python
X=df.iloc[:,0:2]
Y=df['target']
```

```python
clf=Perceptron()
clf.fit(X,Y)
```
