```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential 
from tensorflow.keras.layers import Dense
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
df=pd.read_csv('sin_dataset_1000.csv'
               )
```

```python
df.head()
```

```python
plt.scatter(df['x'],df['sin_x'])
```

```python
model=Sequential()
model.add(Dense(1000, input_dim=1, activation='relu'))
model.add(Dense(500, activation='relu'))
model.add(Dense(250, activation='relu'))
model.add(Dense(125, activation='relu'))
model.add(Dense(8, activation='relu'))
model.add(Dense(1,activation='linear'))
```

```python
model.compile(loss='mse',optimizer='adam',metrics=['mse'])
```

```python
history=model.fit(df[['x']],df['sin_x'],epochs=300,validation_split=0.2,verbose=True)
```

```python
plt.scatter(df['x'],df['sin_x'])
plt.scatter(df['x'],model.predict(df['x']))
```
